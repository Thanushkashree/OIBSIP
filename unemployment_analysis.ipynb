# Unemployment Analysis with Python

## Data Collection & Loading

This section focuses on importing the required Python libraries and loading the unemployment datasets for analysis.


```python
# Task 2: Unemployment Analysis with Python
# Cell 1: Data Collection & Loading – Importing Libraries and Loading Unemployment Datasets

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set(style="whitegrid")

# Load the datasets
unemployment_df = pd.read_csv("Unemployment in India.csv")
unemployment_rate_df = pd.read_csv("Unemployment_Rate_upto_11_2020.csv")

print("Unemployment datasets loaded successfully!")

print("\nUnemployment in India Dataset:")
display(unemployment_df.head())

print("\nUnemployment Rate Dataset:")
display(unemployment_rate_df.head())
```

## Data Understanding & Exploration

This section examines the structure, data types, missing values and statistical characteristics of the unemployment datasets.

```python
# Cell 2: Data Understanding & Exploration – Inspecting Dataset Structure, Data Types, Missing Values and Statistics

print("===== Unemployment in India Dataset =====")

print("\nShape:")
print(unemployment_df.shape)

print("\nColumns:")
print(unemployment_df.columns.tolist())

print("\nData Types:")
print(unemployment_df.dtypes)

print("\nMissing Values:")
print(unemployment_df.isnull().sum())

print("\nDescriptive Statistics:")
display(unemployment_df.describe(include="all"))


print("\n\n===== Unemployment Rate Dataset =====")

print("\nShape:")
print(unemployment_rate_df.shape)

print("\nColumns:")
print(unemployment_rate_df.columns.tolist())

print("\nData Types:")
print(unemployment_rate_df.dtypes)

print("\nMissing Values:")
print(unemployment_rate_df.isnull().sum())

print("\nDescriptive Statistics:")
display(unemployment_rate_df.describe(include="all"))
```

## Data Cleaning & Preparation

This section prepares the datasets for analysis by standardizing column names, converting dates, handling missing values and organizing the data chronologically.

```python
# Cell 3: Data Cleaning & Preparation – Standardizing Columns, Converting Dates and Handling Missing Values

# Remove extra spaces from column names
unemployment_df.columns = unemployment_df.columns.str.strip()
unemployment_rate_df.columns = unemployment_rate_df.columns.str.strip()

# Display the column names after cleaning
print("Unemployment in India columns:")
print(unemployment_df.columns.tolist())

print("\nUnemployment Rate columns:")
print(unemployment_rate_df.columns.tolist())

# Convert Date columns into datetime format
unemployment_df["Date"] = pd.to_datetime(unemployment_df["Date"], errors="coerce")
unemployment_rate_df["Date"] = pd.to_datetime(unemployment_rate_df["Date"], errors="coerce")

# Remove rows containing missing values
unemployment_df = unemployment_df.dropna().copy()
unemployment_rate_df = unemployment_rate_df.dropna().copy()

# Sort data by date
unemployment_df = unemployment_df.sort_values("Date").reset_index(drop=True)
unemployment_rate_df = unemployment_rate_df.sort_values("Date").reset_index(drop=True)

print("\nData cleaning and preparation completed successfully!")

print("\nRemaining missing values:")
print("Unemployment in India:", unemployment_df.isnull().sum().sum())
print("Unemployment Rate:", unemployment_rate_df.isnull().sum().sum())
```

## Exploratory Data Analysis (EDA)

This section analyzes the distribution of unemployment rates and identifies the spread and possible outliers using visualizations.

```python
# Cell 4: Exploratory Data Analysis – Analyzing Unemployment Rate Distribution and Outliers

# Create histogram to understand the distribution of unemployment rates
plt.figure(figsize=(10, 6))

sns.histplot(
    unemployment_df["Estimated Unemployment Rate (%)"],
    bins=20,
    kde=True
)

plt.title("Distribution of Unemployment Rate in India")
plt.xlabel("Estimated Unemployment Rate (%)")
plt.ylabel("Frequency")
plt.show()


# Create boxplot to identify spread and possible outliers
plt.figure(figsize=(12, 6))

sns.boxplot(
    x=unemployment_df["Estimated Unemployment Rate (%)"]
)

plt.title("Distribution and Outliers of Unemployment Rate")
plt.xlabel("Estimated Unemployment Rate (%)")
plt.show()
```

## Regional Analysis

This section compares the average unemployment rates across different states and identifies regions with higher unemployment rates.

```python
# Cell 5: Regional Analysis – Comparing Unemployment Rates Across States

# Calculate average unemployment rate for each state
state_unemployment = (
    unemployment_df.groupby("Region")["Estimated Unemployment Rate (%)"]
    .mean()
    .sort_values(ascending=False)
)

# Display the average unemployment rate of each state
print("Average Unemployment Rate by State:")
display(state_unemployment)


# Select the top 10 states with the highest average unemployment rate
top_10_states = state_unemployment.head(10)

# Create a bar chart for the top 10 states
plt.figure(figsize=(12, 6))

sns.barplot(
    x=top_10_states.values,
    y=top_10_states.index
)

plt.title("Top 10 States with Highest Average Unemployment Rate")
plt.xlabel("Average Unemployment Rate (%)")
plt.ylabel("State")
plt.show()
```

## Time-Series Analysis

This section examines changes in the unemployment rate over time and visualizes the overall unemployment trend.

```python
# Cell 6: Time-Series Analysis – Analyzing Unemployment Trends Over Time

# Calculate the average unemployment rate for each date
monthly_unemployment = (
    unemployment_df.groupby("Date")["Estimated Unemployment Rate (%)"]
    .mean()
    .reset_index()
)

# Create a line chart to visualize unemployment trends over time
plt.figure(figsize=(14, 6))

sns.lineplot(
    data=monthly_unemployment,
    x="Date",
    y="Estimated Unemployment Rate (%)",
    marker="o"
)

plt.title("Unemployment Rate Trend Over Time")
plt.xlabel("Date")
plt.ylabel("Average Unemployment Rate (%)")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

## Insights & Conclusion

This section summarizes the major findings from the unemployment analysis, including overall unemployment levels, regional differences and trends over time.

```python
# Cell 7: Insights & Conclusion – Identifying Major Unemployment Trends

# Calculate overall unemployment statistics
average_unemployment = unemployment_df["Estimated Unemployment Rate (%)"].mean()
maximum_unemployment = unemployment_df["Estimated Unemployment Rate (%)"].max()
minimum_unemployment = unemployment_df["Estimated Unemployment Rate (%)"].min()

# Find the state with the highest average unemployment rate
highest_state = state_unemployment.idxmax()
highest_state_rate = state_unemployment.max()

# Find the state with the lowest average unemployment rate
lowest_state = state_unemployment.idxmin()
lowest_state_rate = state_unemployment.min()

# Display the major findings
print("===== Unemployment Analysis – Key Insights =====")

print(f"\nOverall Average Unemployment Rate: {average_unemployment:.2f}%")

print(f"Highest Recorded Unemployment Rate: {maximum_unemployment:.2f}%")

print(f"Lowest Recorded Unemployment Rate: {minimum_unemployment:.2f}%")

print(f"\nState with Highest Average Unemployment Rate: {highest_state}")
print(f"Average Rate: {highest_state_rate:.2f}%")

print(f"\nState with Lowest Average Unemployment Rate: {lowest_state}")
print(f"Average Rate: {lowest_state_rate:.2f}%")

print("\nThe analysis shows variations in unemployment rates across different states and time periods.")
print("The visualizations help identify unemployment distribution, regional differences and trends over time.")
```
